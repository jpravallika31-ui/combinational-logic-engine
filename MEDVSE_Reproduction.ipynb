```python
import numpy as np
import os

def load_mths_2026(data_path, mode="hr"):
    if mode not in ("hr", "spo2"):
        raise ValueError("mode must be 'hr' or 'spo2'")

    X, y, metadata = [], [], []

    for subject_id in range(67):
        signal_path = os.path.join(
            data_path, f"signal_{subject_id}.npy"
        )
        label_path = os.path.join(
            data_path, f"label_{subject_id}.npy"
        )

        if not os.path.exists(signal_path):
            continue

        if not os.path.exists(label_path):
            raise FileNotFoundError(label_path)

        signal = np.load(signal_path)
        labels = np.load(label_path)

        if signal.ndim != 2 or signal.shape[1] != 3:
            raise ValueError(f"Unexpected signal shape: {subject_id}")

        if labels.ndim != 2 or labels.shape[1] != 2:
            raise ValueError(f"Unexpected label shape: {subject_id}")

        if len(signal) != len(labels) * 30:
            raise ValueError(f"Duration mismatch: {subject_id}")

        complete_windows = min(
            len(signal) // 300,
            len(labels) // 10
        )

        for window_index in range(complete_windows):
            signal_start = window_index * 300
            label_start = window_index * 10

            signal_window = signal[
                signal_start:signal_start + 300
            ]

            label_window = labels[
                label_start:label_start + 10
            ]

            # Exclude windows with invalid or non-finite values.
            if not np.isfinite(signal_window).all():
                continue

            if not np.isfinite(label_window).all():
                continue

            if np.any(label_window == -1):
                continue

            if mode == "hr":
                x_window = signal_window[:, 1:2]
                target = np.mean(label_window[:, 0])

            else:
                x_window = signal_window[:, [0, 2]]
                target = np.mean(label_window[:, 1])

            X.append(x_window)
            y.append(target)

            metadata.append({
                "subject_id": subject_id,
                "window_index": window_index
            })

    return (
        np.asarray(X, dtype=np.float32),
        np.asarray(y, dtype=np.float32),
        metadata
    )
```

```python
subject_fold_map = {}

for i, item in enumerate(meta_hr):
    subject_id = item["subject_id"]
    fold = int(fold_assignments[i])

    subject_fold_map.setdefault(subject_id, set()).add(fold)

invalid_subjects = {
    subject: folds
    for subject, folds in subject_fold_map.items()
    if len(folds) != 1
}

print("Subjects assigned to multiple test folds:", invalid_subjects)
print("Total subjects assigned:", len(subject_fold_map))
```

```python
import os
import glob
import numpy as np

def prepare_mths(data_path):
    X_hr, y_hr = [], []
    X_spo2, y_spo2 = [], []
    metadata = []

    signal_files = sorted(glob.glob(data_path + "/signal_*.npy"))

    for signal_file in signal_files:
        subject_id = os.path.basename(signal_file).replace(
            "signal_", ""
        ).replace(".npy", "")

        label_file = os.path.join(
            data_path, f"label_{subject_id}.npy"
        )

        signal = np.load(signal_file)
        labels = np.load(label_file)

        # Process 10-second non-overlapping windows
        for start in range(0, len(signal) - 300 + 1, 300):
            end = start + 300

            signal_window = signal[start:end]
            label_window = labels[start // 30:(end // 30)]

            # Skip invalid or non-finite windows
            if not np.isfinite(signal_window).all():
                continue

            if not np.isfinite(label_window).all():
                continue

            if np.any(label_window == -1):
                continue

            # HR: Green channel
            X_hr.append(signal_window[:, 1:2])
            y_hr.append(np.mean(label_window[:, 0]))

            # SpO2: Red and Blue channels
            X_spo2.append(signal_window[:, [0, 2]])
            y_spo2.append(np.mean(label_window[:, 1]))

            metadata.append(int(subject_id))

    return (
        np.array(X_hr),
        np.array(y_hr),
        np.array(X_spo2),
        np.array(y_spo2),
        np.array(metadata)
    )
```

```python
fold_plans = []

for fold in range(10):
    test_mask = fold_assignments == fold

    remaining_indices = np.where(~test_mask)[0]
    remaining_groups = metadata[remaining_indices]

    splitter = GroupShuffleSplit(
        n_splits=1,
        test_size=0.15,
        random_state=1400
    )

    train_local, valid_local = next(
        splitter.split(
            remaining_indices,
            groups=remaining_groups
        )
    )

    train_indices = remaining_indices[train_local]
    valid_indices = remaining_indices[valid_local]
    test_indices = np.where(test_mask)[0]

    fold_plans.append({
        "train": train_indices,
        "validation": valid_indices,
        "test": test_indices
    })

    print(
        f"Fold {fold}: "
        f"Train={len(train_indices)}, "
        f"Validation={len(valid_indices)}, "
        f"Test={len(test_indices)}"
    )
```
